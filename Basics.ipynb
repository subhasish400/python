## Indentation

Indentation in python is used to define the structure and hierarchy of the code.Unlike many other programming languages which use braces {} to delimit blocks of code,python uses indentation to determine the grouping of statements.This means that all the statements within a block must be intended at the same level.

```python
# Write a program in python to know about how the indentation works
if True:
    print('Correct Indentation')
    if False:
        print('This out print')
    print('This will print')
print('Outside the if block')
```

```python
# Indentation: Python uses indentation to define blocks of code.Consistant use of spaces (commonly 4) or a tab is required.
age = 24
if age > 30:
    print(age)
print(age)
```

## Operators

An operator is a symbol which performs an operation.There are 6 types of operators are there such as 1. Arithematic operator 2.Relational operator/Comparision operator 3.Logical operator 4.Assignment operator 5.Bitwise operator and special type of operator

Again there are two types of special type of operators are there such as 1. Identity operator and membership operator

Arithematic Operator

Arithematic operators are used to perform basic arithematic operations like addition, substraction, multiplication, division etc

```python
a = 10
b = 5
print("The addition of {} and {} becomes: {}".format(a,b,a+b))
print("The substraction of {} and {} becomes: {}".format(a,b,a-b))
print("The multiplication of {} and {} becomes: {}".format(a,b,a*b))
print("The division of {} and {} becomes: {}".format(a,b,a/b))
print("The modulous of {} and {} becomes: {}".format(a,b,a%b))
print("The floor division of {} and {} becomes: {}".format(a,b,a//b))
print("The power of {} and {} becomes: {}".format(a,b,a**b))
```

Comparision Operator

Relational and comparision operators are used to compare the value of operands(expressions) to produce logical value.Here a logical value is either true or false

```python
a = 10
b = 10
print(a==b)
# Here it returns true because the content stored in both the variable will be same
s1 = 'Subhasish'
s2 = 'subhasish'
print(s1==s2)
# It returns false because python is a case sensitive language
print(s1!=s2)
# greater than
num1 = 10
num2 = 20
print(num1>num2)
# less than
print(num1<num2)
number_1 = 45
number_2 = 45
print(number_1>=number_2)
```

Logical Operators

Logical operators are used to connect more relational operators to form a complex expression called logical expression.A value obtained by a logical operation is always logical ie(either True or False)

```python
# And
x = True
y = True
result = x and y 
print(result)
# Or
x = True
y = False
result = x or y
print(result)
# Not
x = False
print(not x)
```
