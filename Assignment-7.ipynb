```python
class Vector:
    def __init__(self,x,y):
        self.x = x
        self.y = y
    def __add__(self,other):
        return Vector(self.x + other.x, self.y + other.y)
    def __str__(self):
        return f"Vector({self.x}, {self.y})"
v1 = Vector(2, 3)
v2 = Vector(4, 5)
v3 = v1 + v2
print(f'The addition of two vectors becomes:\n{v3}')
```

Create a custom exception named `InsufficientBalanceError`. In the `BankAccount` class, raise this exception when a withdrawal amount is greater than the balance. Handle the exception and print an appropriate message.

```python
class InsufficientBalanceError(Exception):
    pass
class BankAccount:
    def __init__(self,account_number,balance = 0):
        self.account_number = account_number
        self.balance = balance
    def deposit(self,amount):
        self.balance = self.balance + amount
        return f'After deposite the balance of the person having the account number {account_number} becomes: {self.balance}'
    def withdraw(self,amount):
        try:
            if amount > self.balance:
                raise InsufficientBalanceError('You have Insufficient Balance!')
        except InsufficientBalanceError as e:
            return f'{e}'
        else:
            self.balance = self.balance - amount
            return f'after withdraw the balance of the person having the account number {account_number} becomes: {self.balance}'
    def check_balance(self):
        return f'The balance of the person having the account number {account_number} becomes: {self.balance}'
account_number = eval(input('Enter the account number:'))
account = BankAccount(account_number, 1000)
print(account.check_balance())
amount = eval(input('Enter the amount you want to deposite:'))
print(account.deposit(amount))
amount = eval(input('Enter the amount you want to withdraw:'))
print(account.withdraw(amount))
print(account.check_balance())
amount = eval(input('Enter the amount you want to withdraw:'))
print(account.withdraw(amount))
```

Create a class named `FileManager` that implements the context manager protocol to open and close a file. Use this class to read the contents of a file.

```python
class FileManager:
    def __init__(self,filename,mode):
        self.filename = filename
        self.mode = mode
    def open_file(self):
        with open(self.filename,self.mode) as self.file:
            content = self.file.read()
            print(content)
    def close_file(self):
        self.file.close()
file = FileManager('Source.txt','r')
file.open_file()
file.close_file()
```

Create a class named `Calculator` with methods to add, subtract, multiply, and divide. Each method should return the object itself to allow method chaining. Create an object and chain multiple method calls.

```python
class Calculator:
    def __init__(self,value=0):
        self.value = value
    def add(self,amount):
        self.value = self.value + amount
        return self
    def substract(self,amount):
        self.value = self.value - amount
        return self
    def multiply(self,amount):
        self.value = self.value * amount
        return self 
    def divide(self,amount):
        if amount != 0:
            self.value = self.value // amount
            return self 
        else:
            return f'Division by zero is not possible'
calc = Calculator()
calc.add(10).substract(3).multiply(2).divide(2)
print(calc.value)  
            
    
        
```
